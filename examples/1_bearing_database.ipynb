# Bearing Database

Initialization of `BearingDatabase` include small database of different bearings, with measured internal geometry and validated fault-characteristic-frequencies. Database is stored inside pandas `DataFrame`.

```python
from pybearing.data.bearing_database import BearingDatabase

database = BearingDatabase()

database.database
```

One can set its own database with `set_database` method. As argument it takes `DataFrame`, structured in a way to not break other components of the `pybearing` library, therefore `check_columns` from `pybearing.data.utils` should be used beforehand.

```python
import pandas as pd
from pybearing.data.utils import check_columns

new_database = pd.DataFrame(columns=["name", "manufacturer", "d", "D", "B", "D_b", "D_m", "theta", "N_b", "f_rot", "f_c", "f_re_ax", "f_o", "f_re", "f_i"])

print(f"New database is valid: {check_columns(new_database)}")

database.set_database(new_database)
```

Database can also be exported to a CSV file using the `export_database` method.

```python
database.export_database(r"path_to_folder", r"database_name")
```

Currently database enables search for different parameters by name. If something is found in database, it return us `Bearing` or `FaultFrequencies` objects or list of them if multiple are found.

```python
database.find_bearing_by_name("NMB_1560kk")
```

```python
database.find_characteristic_frequencies_by_bearing_name("NMB_1560kk")
```
